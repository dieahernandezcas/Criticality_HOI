# Figure 0 — Methodology Overview

Generates four panels (A–D) for the methodology figure of the paper:
- **Panel A**: Research question
- **Panel B**: Wilson–Cowan model and Taylor expansion
- **Panel C**: Four coupling architectures
- **Panel D**: Parameter sweeps and measures

Output: SVG files saved to `results/`

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Arc, Polygon
from matplotlib.lines import Line2D
import numpy as np
import os

# ── Global style ──
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Latin Modern Roman', 'STIXGeneral', 'DejaVu Serif'],
    'mathtext.fontset': 'cm',
    'font.size': 9,
    'axes.linewidth': 0.6,
    'svg.fonttype': 'none',  # editable text in SVG
})

# ── Pastel academic palette ──
C_BLUE    = '#A8C8E8'   # soft blue
C_CORAL   = '#E8A8A8'   # soft coral/red
C_GREEN   = '#A8D8B0'   # soft green
C_PURPLE  = '#C4A8D8'   # soft purple
C_YELLOW  = '#E8D8A0'   # soft gold
C_GRAY    = '#D0D0D0'   # light gray
C_DARK    = '#3C3C3C'   # dark text
C_BORDER  = '#707070'   # border color

OUTDIR = os.path.join(os.path.dirname(os.path.abspath('.')), 'results')
os.makedirs(OUTDIR, exist_ok=True)

print(f'Output directory: {OUTDIR}')
```

```python
# ════════════════════════════════════════════════════════════
# PANEL A — High-order interactions problem (OPTIMIZED FOR 2-COLUMN LAYOUT)
# ════════════════════════════════════════════════════════════

fig = plt.figure(figsize=(4.0, 3.0))  # Optimized for 2-column revtex
ax_net = fig.add_axes([0.05, 0.45, 0.4, 0.5])
ax_signals = fig.add_axes([0.5, 0.35, 0.48, 0.6])
ax_question = fig.add_axes([0.05, 0.05, 0.9, 0.25])

# ── Left: Networks with UNDIRECTED curved lines (no arrows) ──
ax_net.set_xlim(-1.5, 3.5)
ax_net.set_ylim(-0.5, 3)
ax_net.axis('off')
ax_net.set_aspect('equal')

# Helper: draw UNDIRECTED curved line between two points (Bezier curve)
def draw_curved_line(ax, x1, y1, x2, y2, color=C_DARK, lw=1.0, rad=0.25):
    """Draw a smooth undirected curved line from (x1,y1) to (x2,y2)."""
    t_curve = np.linspace(0, 1, 100)
    # Calculate perpendicular offset for control point
    dx, dy = x2 - x1, y2 - y1
    dist = np.sqrt(dx**2 + dy**2)
    perp_x = -dy / dist if dist > 0 else 0
    perp_y = dx / dist if dist > 0 else 0
    ctrl_x = (x1 + x2) / 2 + rad * perp_x
    ctrl_y = (y1 + y2) / 2 + rad * perp_y
    # Quadratic Bezier curve
    x_curve = (1-t_curve)**2 * x1 + 2*(1-t_curve)*t_curve * ctrl_x + t_curve**2 * x2
    y_curve = (1-t_curve)**2 * y1 + 2*(1-t_curve)*t_curve * ctrl_y + t_curve**2 * y2
    ax.plot(x_curve, y_curve, color=color, lw=lw, zorder=2)

# 2nd-order network (left)
ax_net.text(0.5, 2.7, '2nd-order', ha='center', fontsize=8, fontweight='bold', color=C_DARK)
pos_2nd = [(0.2, 2.0), (1.0, 1.2), (0.2, 0.4)]
for x, y in pos_2nd:
    ax_net.add_patch(plt.Circle((x, y), 0.25, fc=C_BLUE, ec=C_BORDER, lw=1.0, zorder=5))
for i, (x1, y1) in enumerate(pos_2nd):
    ax_net.text(x1, y1, str(i+1), ha='center', va='center', fontsize=8, fontweight='bold', color=C_DARK, zorder=6)

# Undirected curved lines for 2nd-order (one per pair)
draw_curved_line(ax_net, 0.2, 2.0, 1.0, 1.2, color=C_BORDER, lw=1.0, rad=0.25)  # 1-2
draw_curved_line(ax_net, 1.0, 1.2, 0.2, 0.4, color=C_BORDER, lw=1.0, rad=0.25)  # 2-3
draw_curved_line(ax_net, 0.2, 0.4, 0.2, 2.0, color=C_BORDER, lw=1.0, rad=-0.4)  # 3-1

# 3rd-order network (right)
ax_net.text(2.5, 2.7, '3rd-order', ha='center', fontsize=8, fontweight='bold', color=C_DARK)
pos_3rd = [(2.2, 2.0), (3.0, 1.2), (2.2, 0.4)]
for x, y in pos_3rd:
    ax_net.add_patch(plt.Circle((x, y), 0.25, fc=C_CORAL, ec=C_BORDER, lw=1.0, zorder=5))
for i, (x, y) in enumerate(pos_3rd):
    ax_net.text(x, y, str(i+1), ha='center', va='center', fontsize=8, fontweight='bold', color=C_DARK, zorder=6)

# Hyperedge triangle for 3rd-order
tri = Polygon(pos_3rd, closed=True, fc=C_CORAL, ec='none', alpha=0.15, zorder=1)
ax_net.add_patch(tri)

# Same undirected curved lines
draw_curved_line(ax_net, 2.2, 2.0, 3.0, 1.2, color=C_BORDER, lw=1.0, rad=0.25)  # 1-2
draw_curved_line(ax_net, 3.0, 1.2, 2.2, 0.4, color=C_BORDER, lw=1.0, rad=0.25)  # 2-3
draw_curved_line(ax_net, 2.2, 0.4, 2.2, 2.0, color=C_BORDER, lw=1.0, rad=-0.4)  # 3-1

# Question: 2nd-order vs 3rd-order?
ax_net.text(1.3, 1.2, '=', ha='center', fontsize=14, fontweight='bold', color='#B05050')

# ── Right: Time series for each node with DIFFERENT COLORS and FREQUENCIES ──
ax_signals.set_xlim(0, 10)
ax_signals.set_ylim(-0.5, 3.5)
ax_signals.axis('off')

# Define colors for each node signal
color_signal1 = '#4A7BA7'  # darker blue for signal 1
color_signal2 = '#9B6B5E'  # brownish for signal 2
color_signal3 = '#6BA876'  # green for signal 3

# Generate oscillatory signals with DISTINCTLY DIFFERENT FREQUENCIES
t = np.linspace(0, 4*np.pi, 200)

# Signal 1 (Node 1): frequency f₁ = 2.0 Hz
f1 = 2.0
y1_raw = np.sin(f1*t)
y1 = 0.5 + 0.35 * y1_raw

# Signal 2 (Node 2): frequency f₂ = 1.4 Hz (slower)
f2 = 1.4
y2_raw = np.sin(f2*t + 0.5)
y2 = 1.5 + 0.35 * y2_raw

# Signal 3 (Node 3): frequency f₃ = 2.7 Hz (faster)
f3 = 2.7
y3_raw = np.sin(f3*t + 0.8)
y3 = 2.5 + 0.35 * y3_raw

# Scale time to fit axes
t_scaled = t / (4*np.pi) * 9 + 0.5

# Plot signals with different colors
ax_signals.plot(t_scaled, y1, color=color_signal1, lw=1.5, zorder=3)
ax_signals.plot(t_scaled, y2, color=color_signal2, lw=1.5, zorder=3)
ax_signals.plot(t_scaled, y3, color=color_signal3, lw=1.5, zorder=3)

# Labels on the left with node colors
ax_signals.text(0.05, 0.5, 'Signal 1', fontsize=7, color=color_signal1, fontweight='bold')
ax_signals.text(0.05, 1.5, 'Signal 2', fontsize=7, color=color_signal2, fontweight='bold')
ax_signals.text(0.05, 2.5, 'Signal 3', fontsize=7, color=color_signal3, fontweight='bold')

ax_signals.text(5.0, 3.2, 'Same pattern,\ndifferent frequencies', ha='center', fontsize=6.5, color=C_DARK, 
               bbox=dict(boxstyle='round,pad=0.3', fc=C_YELLOW, alpha=0.4, ec=C_BORDER, lw=0.5))

# ── Bottom: Question ──
ax_question.text(0.5, 0.6, 'Can 2nd-order interactions generate high-order statistical dependencies?',
                ha='center', va='center', fontsize=8, fontweight='bold', color=C_DARK,
                transform=ax_question.transAxes,
                bbox=dict(boxstyle='round,pad=0.5', fc=C_YELLOW, alpha=0.3, ec=C_BORDER, lw=0.8))
ax_question.axis('off')

# Panel label
fig.text(0.02, 0.95, 'A', fontsize=14, fontweight='bold', color=C_DARK)

fig.savefig(os.path.join(OUTDIR, 'fig0_panelA.svg'), format='svg',
            transparent=True, bbox_inches='tight', dpi=300)
plt.close()
print('Panel A saved (4.0\" × 3.0\")')
```

```python
# ════════════════════════════════════════════════════════════
# PANEL B — Poincaré sections (OPTIMIZED FOR 2-COLUMN LAYOUT)
# ════════════════════════════════════════════════════════════

fig, axes = plt.subplots(1, 3, figsize=(4.0, 1.8))

# Generate phase space trajectories with noise
t_phase = np.linspace(0, 50, 1000)
np.random.seed(42)

for idx, ax_p in enumerate(axes):
    ax_p.set_xlim(-0.2, 1.2)
    ax_p.set_ylim(-0.2, 1.2)
    ax_p.set_xlabel(r'$E_i$', fontsize=7)
    ax_p.set_ylabel(r'$I_i$', fontsize=7)
    ax_p.set_title(f'Node {idx+1}', fontsize=7.5, fontweight='bold', color=C_DARK)
    ax_p.tick_params(labelsize=6)
    ax_p.spines['top'].set_visible(False)
    ax_p.spines['right'].set_visible(False)

    if idx < 2:  # Oscillatory for nodes 1 and 2 (WITH NOISE)
        # Base oscillatory trajectory (limit cycle)
        E_osc = 0.5 + 0.3 * np.cos(t_phase + idx*np.pi/6)
        I_osc = 0.5 + 0.25 * np.sin(t_phase + idx*np.pi/6)
        
        # Add small noise
        noise_amplitude = 0.02
        E_osc = E_osc + noise_amplitude * np.random.normal(0, 1, len(E_osc))
        I_osc = I_osc + noise_amplitude * np.random.normal(0, 1, len(I_osc))

        ax_p.plot(E_osc, I_osc, color=C_BLUE, lw=0.8, zorder=3, alpha=0.8)
        
        # Add arrow to show direction
        ax_p.annotate('', xy=(E_osc[250], I_osc[250]), xytext=(E_osc[200], I_osc[200]),
                     arrowprops=dict(arrowstyle='->', color=C_BLUE, lw=0.8))
        
        # Mark starting point
        ax_p.plot(E_osc[0], I_osc[0], 'o', color=C_BLUE, markersize=4, zorder=5)
        
        # Annotation
        ax_p.text(0.5, -0.1, 'Osc', ha='center', fontsize=6.5, color=C_BLUE, fontweight='bold')
    
    else:  # Fixed point for node 3 (NO NOISE)
        # Trajectory spiraling to fixed point
        E_fp = 0.6 + 0.4 * np.exp(-0.1*t_phase) * np.cos(t_phase)
        I_fp = 0.6 + 0.35 * np.exp(-0.1*t_phase) * np.sin(t_phase)
        ax_p.plot(E_fp, I_fp, color=C_CORAL, lw=0.8, zorder=3)
        
        # Mark fixed point
        ax_p.plot(0.6, 0.6, 'X', color=C_CORAL, markersize=8, markeredgewidth=1.2, zorder=5)
        
        # Mark starting point
        ax_p.plot(E_fp[0], I_fp[0], 'o', color=C_CORAL, markersize=4, zorder=5)
        
        # Arrow showing convergence
        ax_p.annotate('', xy=(0.62, 0.62), xytext=(E_fp[100], I_fp[100]),
                     arrowprops=dict(arrowstyle='->', color=C_CORAL, lw=0.8))
        
        # Annotation
        ax_p.text(0.5, -0.1, 'FP', ha='center', fontsize=6.5, color=C_CORAL, fontweight='bold')

# Panel label
fig.text(0.02, 0.95, 'B', fontsize=14, fontweight='bold', color=C_DARK)

fig.suptitle('Poincaré sections', fontsize=8, fontweight='bold', color=C_DARK, y=0.98)

plt.tight_layout()
fig.savefig(os.path.join(OUTDIR, 'fig0_panelB.svg'), format='svg',
            transparent=True, bbox_inches='tight', dpi=300)
plt.close()
print('Panel B saved (4.0\" × 1.8\")')
```

```python
# ════════════════════════════════════════════════════════════
# PANEL C — Four coupling models (OPTIMIZED FOR 2-COLUMN LAYOUT)
# ════════════════════════════════════════════════════════════

fig, axes = plt.subplots(2, 2, figsize=(4.0, 3.5))

model_info = [
    ('2nd-order\nadditive', C_BLUE, False, False),
    ('2nd-order\ndiffusive', C_BLUE, False, True),
    ('3rd-order\nadditive', C_CORAL, True, False),
    ('3rd-order\ndiffusive', C_CORAL, True, True),
]

positions_template = [(0.3, 0.7), (0.8, 0.5), (0.3, 0.2)]

for (ax_model, (label, color, has_hyper, is_diff)) in zip(axes.flat, model_info):
    ax_model.set_xlim(0, 1)
    ax_model.set_ylim(0, 1)
    ax_model.axis('off')
    ax_model.set_aspect('equal')
    
    # Title
    ax_model.text(0.5, 0.95, label, ha='center', va='top', fontsize=7.5, fontweight='bold', color=C_DARK)
    
    # Positions
    pos = positions_template
    r_node = 0.06
    
    # Hyperedge triangle (if 3rd-order)
    if has_hyper:
        tri = Polygon(pos, closed=True, fc=color, ec='none', alpha=0.15, zorder=1)
        ax_model.add_patch(tri)
    
    # Curved edges
    for i in range(3):
        for j in range(i+1, 3):
            x1, y1 = pos[i]
            x2, y2 = pos[j]
            # Curved connection
            if is_diff:
                ls = '--'
            else:
                ls = '-'
            
            # Draw curved connection using arc
            t_arc = np.linspace(0, 1, 100)
            # Quadratic Bezier curve
            ctrl_x, ctrl_y = 0.5, 0.5  # Control point
            x_arc = (1-t_arc)**2 * x1 + 2*(1-t_arc)*t_arc * ctrl_x + t_arc**2 * x2
            y_arc = (1-t_arc)**2 * y1 + 2*(1-t_arc)*t_arc * ctrl_y + t_arc**2 * y2
            
            ax_model.plot(x_arc, y_arc, ls, color=C_BORDER, lw=0.9, zorder=2)
    
    # Nodes
    for k, (x, y) in enumerate(pos):
        ax_model.add_patch(plt.Circle((x, y), r_node, fc=color, ec=C_BORDER, lw=0.7, zorder=5))
        ax_model.text(x, y, str(k+1), ha='center', va='center', fontsize=6.5, fontweight='bold',
                     color=C_DARK, zorder=6)
    
    # Type label at bottom
    type_label = 'diffusive' if is_diff else 'additive'
    ax_model.text(0.5, 0.02, type_label, ha='center', fontsize=6.5, color=C_DARK, style='italic')

# Panel label
fig.text(0.02, 0.98, 'C', fontsize=14, fontweight='bold', color=C_DARK)

plt.tight_layout()
fig.savefig(os.path.join(OUTDIR, 'fig0_panelC.svg'), format='svg',
            transparent=True, bbox_inches='tight', dpi=300)
plt.close()
print('Panel C saved (4.0\" × 3.5\")')
```

```python
# ════════════════════════════════════════════════════════════
# PANEL D — Parameter heatmaps (OPTIMIZED FOR 2-COLUMN LAYOUT)
# ════════════════════════════════════════════════════════════

fig, axes = plt.subplots(1, 2, figsize=(4.0, 2.0))

# Create heatmap data with clear separation between FP (low values) and OSC (high values)
K_vals = np.linspace(0, 1, 100)

# Sweep 1: P vs K
P_vals = np.linspace(0, 1, 100)
heatmap1 = np.zeros((len(P_vals), len(K_vals)))

# Fill heatmap1: low values (near 0) below Hopf, high values above Hopf
for i, P in enumerate(P_vals):
    for j, K in enumerate(K_vals):
        P_boundary = 0.2 + 0.6 * np.exp(-3*K)
        if P < P_boundary:
            heatmap1[i, j] = np.random.uniform(0, 0.05)
        else:
            heatmap1[i, j] = 0.65 + 0.15 * np.sin(np.pi * (P - P_boundary) / (1 - P_boundary))

# Sweep 2: aE vs K
aE_vals = np.linspace(0, 1, 100)
heatmap2 = np.zeros((len(aE_vals), len(K_vals)))

# Fill heatmap2
for i, aE in enumerate(aE_vals):
    for j, K in enumerate(K_vals):
        aE_boundary = 0.3 + 0.6 * (1 - np.exp(-2.5*K))
        if aE < aE_boundary:
            heatmap2[i, j] = np.random.uniform(0, 0.05)
        else:
            heatmap2[i, j] = 0.65 + 0.15 * np.sin(np.pi * (aE - aE_boundary) / (1 - aE_boundary))

# Plot heatmaps
im1 = axes[0].contourf(K_vals, P_vals, heatmap1, levels=12, cmap='YlOrRd', alpha=0.8)
im2 = axes[1].contourf(K_vals, aE_vals, heatmap2, levels=12, cmap='YlOrRd', alpha=0.8)

# Hopf boundaries
K_hopf1_plot = np.linspace(0, 1, 100)
P_hopf1_plot = 0.2 + 0.6 * np.exp(-3*K_hopf1_plot)

K_hopf2_plot = np.linspace(0, 1, 100)
aE_hopf2_plot = 0.3 + 0.6 * (1 - np.exp(-2.5*K_hopf2_plot))

# Plot Hopf boundaries
axes[0].plot(K_hopf1_plot, P_hopf1_plot, '--', color='#B05050', lw=1.5, label='Criticality', zorder=5)
axes[1].plot(K_hopf2_plot, aE_hopf2_plot, '--', color='#B05050', lw=1.5, label='Criticality', zorder=5)

# Labels and formatting
axes[0].set_xlabel(r'$K$', fontsize=7.5)
axes[0].set_ylabel(r'$P$', fontsize=7.5)
axes[0].set_title(r'$(P, K)$', fontsize=7.5, fontweight='bold', color=C_DARK)
axes[0].tick_params(labelsize=6)

axes[1].set_xlabel(r'$K$', fontsize=7.5)
axes[1].set_ylabel(r'$a_E$', fontsize=7.5)
axes[1].set_title(r'$(a_E, K)$', fontsize=7.5, fontweight='bold', color=C_DARK)
axes[1].tick_params(labelsize=6)

# Colorbars
cbar1 = plt.colorbar(im1, ax=axes[0], fraction=0.046, pad=0.04)
cbar1.set_label('o-info', fontsize=6.5)
cbar1.ax.tick_params(labelsize=5)

cbar2 = plt.colorbar(im2, ax=axes[1], fraction=0.046, pad=0.04)
cbar2.set_label('o-info', fontsize=6.5)
cbar2.ax.tick_params(labelsize=5)

# Annotations
axes[0].text(0.3, 0.15, 'FP\n≈0', ha='center', fontsize=6, color='white', 
            bbox=dict(boxstyle='round', fc='#333333', alpha=0.7))
axes[0].text(0.8, 0.85, 'OSC\nhigh', ha='center', fontsize=6, color='white',
            bbox=dict(boxstyle='round', fc='#333333', alpha=0.7))

axes[1].text(0.3, 0.15, 'FP\n≈0', ha='center', fontsize=6, color='white',
            bbox=dict(boxstyle='round', fc='#333333', alpha=0.7))
axes[1].text(0.8, 0.85, 'OSC\nhigh', ha='center', fontsize=6, color='white',
            bbox=dict(boxstyle='round', fc='#333333', alpha=0.7))

# Panel label
fig.text(0.02, 0.98, 'D', fontsize=14, fontweight='bold', color=C_DARK)

plt.tight_layout(rect=[0, 0, 1, 0.95])
fig.savefig(os.path.join(OUTDIR, 'fig0_panelD.svg'), format='svg',
            transparent=True, bbox_inches='tight', dpi=300)
plt.close()
print('Panel D saved (4.0\" × 2.0\")')
```
